Imported the main libraries that will be used for data analysis

created a new dataframe that contains only numerical variables using the select_dtypes() function. This allows numerical variables to be analyzed separately.

```python
numerical_df = df.select_dtypes(include=['int64','float64'])
numerical_df.head()
```

A loop was used to generate histograms for each numerical variable

```python
for column in numerical_df.columns:
    plt.figure()
    numerical_df[column].hist()
    plt.title(column)
    plt.show()
```

A loop was used to generate kernel density plots for each numerical variable

```python
for column in numerical_df.columns:
    plt.figure()
    sns.kdeplot(numerical_df[column], fill=True)
    plt.title(column)
    plt.show()
```

A loop was used to generate boxplots for each numerical variable to identify outlies

```python
for column in numerical_df.columns:
  plt.figure()
  sns.boxplot(x=numerical_df[column])
  plt.title(column)
  plt.show()
```

created a new dataframe that contains only categorical variables using the select_dtypes() function. This allows numerical variables to be analyzed separately.

```python
categorical_df = df.select_dtypes(include=['object'])
categorical_df.head()
```

```python
for column in categorical_df.columns:
    plt.figure()
    categorical_df[column].value_counts().plot(kind='bar')
    plt.title(column)
    plt.show()
```

Exploring the relationship between numerical variables and the target variable (Response) using correlation matrices

```python
corr_matrix = numerical_df.corr()
plt.figure(figsize=(12,8))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix of Numerical Variables")
plt.show()
```

Exploring the relationship between categorical variables and the target variable using bar plots

```python
for column in categorical_df.columns:
 plt.figure(figsize=(6,4))
 sns.countplot(x=column, hue="Response", data=df)
 plt.title(f"{column} vs Response")
 plt.xticks(rotation=45)
 plt.show()
```

Exploring the relationship between numerical and categorical variables using box plots 

```python
for cat_col in categorical_df.columns:
    for num_col in numerical_df.columns:
        plt.figure(figsize=(6,4))
        sns.boxplot(x=df[cat_col], y=df[num_col])
        plt.title(f"{num_col} vs {cat_col}")
        plt.xticks(rotation=45)
        plt.show()
```
